```python
IMG_DIR = r"D:\Modern Route-Full Stack GenerativeAI And Agentic AI  code\25-Multimodel-Finetuning\iphone_imgs"

image_files = sorted([
    os.path.join(IMG_DIR, f)
    for f in os.listdir(IMG_DIR)
    if f.lower().endswith((".jpg", ".jpeg", ".png", ".webp"))
])

print("Total images:", len(image_files))
```

```python
captions = [
    "A orange iPhone shown from the back on a white background.",
    "Three iPhones in white, orange, and dark blue shown together from the back.",
    "An orange iPhone shown from the back and front on a white background.",
    "A dark blue iPhone shown from the back with a side view on a transparent background.",
    "An white iPhone shown from the back on a transparent background.",
]
```

```python
instruction = "Describe this iPhone product image in one sentence."
```

```python
rows = []
```

```python
[{

}]
```

```python
for img_path, cap in zip(image_files, captions):
  print(img_path, cap)
```

```python
for img_path, cap in zip(image_files, captions):
    rows.append({
        "image": img_path,
        "text": cap,
        "messages": [
            {
                "role": "user",
                "content": [
                    {"type": "text", "text": instruction},
                    {"type": "image", "image": img_path},
                ],
            },
            {
                "role": "assistant",
                "content": [
                    {"type": "text", "text": cap},
                ],
            },
        ],
    })
```
